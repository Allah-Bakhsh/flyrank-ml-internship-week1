## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (e.g. "flag pages older than 365 days") already failed an honesty check in Week 1, pages trending down weren't reliably the older ones; pages trending up/stable were actually slightly older on average. A single hand-written threshold can't capture that kind of counter-intuitive interaction. A model can combine several weak, non-obvious signals (age, position, CTR, impressions, trend) simultaneously and weigh how they interact  
something a fixed if/else rule structurally can't do.
