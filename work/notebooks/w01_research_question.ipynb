## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*


**Question:** Given a page's observable signals (age, position, CTR, impressions, word count, trend direction), which pages are the best candidates for a content refresh review this cycle?

**Unit of analysis:** one row = one content page.

**Decision it improves:** which pages a limited capacity content review team looks at first.

**Action someone takes:** a human reviewer opens the top-ranked pages and decides whether to rewrite, update, consolidate, or leave alone  the model doesn't refresh anything itself, it orders the queue.

**Cost of a wrong call:** a false positive wastes reviewer time on a page that didn't need attention; a false negative means a genuinely declining or stale page keeps losing traffic/rankings unnoticed until the next cycle.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Loaded the starter dataset below.
Two real numbers: median content age across all pages, and average content age broken down by trend direction, both support treating "how old/stale is this page" as a meaningful signal for the refresh queue.

```python
!wget -q https://raw.githubusercontent.com/Allah-Bakhsh/flyrank-ml-internship-week1/main/data/raw/content_refresh_anonymized.csv -O content_refresh_anonymized.csv
print("Downloaded.")
```

```python
import pandas as pd

df = pd.read_csv("content_refresh_anonymized.csv")

print("Median content_age_days:", df["content_age_days"].median())
print("Pages older than 365 days:", (df["content_age_days"] > 365).mean().round(3), "of all pages")

print("\nAvg content_age_days by trend_direction:")
print(df.groupby("trend_direction")["content_age_days"].mean())

print("\nHand-rule Precision@50 (from Notebook 1/2 baseline): ~0.24")
print("Decision tree Precision@50 (depth=3): ~0.72")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

It **cannot** claim this proves *why* Google ranks pages the way it does, and it cannot claim a refresh *causes* recovery that would need an actual before/after experiment, not just a ranking model.
